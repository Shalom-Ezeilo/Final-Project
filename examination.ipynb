```python
df["city"] = df["city"].str.title()
df["plan"] = df["plan"].str.title()
df["gender"] = df["gender"].str.title()
```

```python
# Standardise existing customer dataset
df["city"] = df["city"].str.strip().str.title()
df["gender"] = df["gender"].str.strip().str.title()
df["plan"] = df["plan"].str.strip().str.title()

```

```python
# Convert monthly_charge to numeric
df["monthly_charge"] = pd.to_numeric(
    df["monthly_charge"].astype(str).str.replace(r"[N,]", "", regex=True),
    errors="coerce"
)
# Convert signup_date to datetime
df["signup_date"] = pd.to_datetime(df["signup_date"], errors="coerce", dayfirst=True)
# Convert churned to 1 or 0
df["churned"] = df["churned"].replace({
    "Yes": 1, "No": 0,
    "yes": 1, "no": 0,
    True: 1, False: 0})

df["churned"] = pd.to_numeric(df["churned"], errors="coerce")
```

```python

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Calculate churn rate by city
city_churn = df.groupby("city")["churned"].mean() * 100

# Create bar chart
plt.figure(figsize=(8, 5))
city_churn.plot(kind="bar", color="skyblue")

plt.title("Churn Rate by City")
plt.xlabel("City")
plt.ylabel("Churn Rate (%)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# Observation
print("Observation: The city with the highest churn rate has the greatest proportion of customers leaving the service.")
```
